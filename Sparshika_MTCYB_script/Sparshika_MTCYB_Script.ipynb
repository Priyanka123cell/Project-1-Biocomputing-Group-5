Build the genetic code as a lookup structure of your own making, then use it to translate the entire coding 
sequence to protein. Report the first 30 residues of your translation and total length. 
You must construct the lookup yourself. You may read the codon assignments from a published table, but 
the data structure and the translation loop must be your own work. 
Hint: A dictionary keyed on the three-letter codon is the natural structure — Lab Session 5, Exercise 4.4. Sixty
four entries written out in full is perfectly acceptable and easier to check than anything clever. 

```python
Table = {
    "TTT":"F", "TCT":"S", "TAT":"Y", "TGT":"C",
    "TTC":"F", "TCC":"S", "TAC":"Y", "TGC":"C",
    "TTA":"L", "TCA":"S", "TAA":"", "TGA":"W",
    "TTG":"L", "TCG":"S", "TAG":"", "TGG":"W",
    "CTT":"L", "CCT":"P", "CAT":"H", "CGT":"R",
    "CTC":"L", "CCC":"P", "CAC":"H", "CGC":"R",
    "CTA":"L", "CCA":"P", "CAA":"Q", "CGA":"R",
    "CTG":"L", "CCG":"P", "CAG":"Q", "CGC":"R",
    "ATT":"I", "ACT":"T", "AAT":"N", "AGT":"S",
    "ATC":"I", "ACC":"T", "AAC":"N", "AGC":"S",
    "ATA":"M", "ACA":"T", "AAA":"K", "AGA":"",
    "ATG":"M", "ACG":"T", "AAG":"K", "AGG":"",
    "GTT":"V", "GCT":"A", "GAT":"D", "GGT":"G",
    "GTC":"V", "GCC":"A", "GAC":"D", "GGC":"G",
    "GTA":"V", "GCA":"A", "GAA":"E", "GGA":"G",
    "GTG":"V", "GCG":"A", "GAG":"E", "GGG":"G"
}
with open("MTCYB_CDS_corrected.fasta", "r") as fh: # open() openes the fasta file in the read mode and fh is a name assigned to this file
    seq = fh.read().replace("\n", "") # Reading the entire file as a continuous string by removing the white lines
Translated_protein_seq = "" # Creating an empty string to add amino acid residues to it
for i in range(0, len(seq), 3): # A for loop is initiated with index 0 to maximum index being the length of the sequence and number of steps being 3 as each codon consists of 3 bases
    codon = seq[i:i+3] # 3 nucleotides from seq is taken and assigned to a variable named codon which acts as a key in the dictionary
    AA_residue = Table[codon] #The key is used to extract the values from table and assign it to a variable named AA_residue
    Translated_protein_seq = Translated_protein_seq + AA_residue # Amino acids are added to the protein sequence 
print (Translated_protein_seq)
print("First 30 residues:", Translated_protein_seq[:30]) # Printing the first 30 amino acid residues
print("Total length of the protein sequence:", len(Translated_protein_seq)) # Calculating the total number of amino acids present in the protein sequence
```
