2(a). Using the CDS coordinates from 1(b), extract the coding sequence from the mRNA. Report its length in 
nucleotides and confirm that it is divisible by three. If it is not, stop and work out why before continuing  
do not proceed with a sequence that fails this check.

```python
with open("refseq_mrna.fasta","r") as refseq:  #to open refseq mrna fasta file and represent it as the variable refseq
    mrna =""   # initializing variable mrna
    for seq in refseq:  # for loop to read each line in refseq
        if(seq[0]!=">"):  # checks if the first line has header symbol not to consider it instead other lines 
            mrna = mrna+seq.strip()  # adds each line and removes whitespace and stored in variable mrna
cds = mrna[354:1686]   #slice from the 355th base pair to the 1686th base pair which is my coding sequence
print("Extracted coding sequence") # prints the text
print(cds)  # prints the extracted coding sequence
print("Length of my extracted coding sequence:",len(cds))  # print length of cds
print("Number of codons in extracted coding sequence:",len(cds)/3)  # checks if the total length of cds is divisible by 3 or not and in this case it is divisible means extraction is correctly done
```

2.(b) Build the genetic code as a lookup structure of your own making, then use it to translate the entire coding 
sequence to protein. Report the first 30 residues of your translation and total length 
You must construct the lookup yourself. You may read the codon assignments from a published table, but
the data structure and the translation loop must be your own work.

```python
codon_table = {"TTT":"F","TTC":"F","TTA":"L","TTG":"L","CTT":"L","CTC":"L","CTA":"L","CTG":"L","ATT":"I","ATC":"I","ATA":"I","ATG":"M","GTT":"V","GTC":"V","GTA":"V","GTG":"V","TCT":"S","TCC":"S","TCA":"S","TCG":"S","CCT":"P","CCC":"P","CCA":"P","CCG":"P","ACT":"T","ACC":"T","ACA":"T","ACG":"T","GCT":"A","GCC":"A","GCA":"A","GCG":"A","TAT":"Y","TAC":"Y","TAA":"*","TAG":"*","CAT":"H","CAC":"H","CAA":"Q","CAG":"Q","AAT":"N","AAC":"N","AAA":"K","AAG":"K","GAT":"D","GAC":"D","GAA":"E","GAG":"E","TGT":"C","TGC":"C","TGA":"*","TGG":"W","CGT":"R","CGC":"R","CGA":"R","CGG":"R","AGT":"S","AGC":"S","AGA":"R","AGG":"R","GGT":"G","GGC":"G","GGA":"G","GGG":"G"}
print("Codon table")
for codon,aa in codon_table.items(): # for loop to print the codon table dictionary  
    print(codon,":",aa)  # prints codon table from the dictionary codon_table
```

```python
amino_seq =""  #initializing variable amino_seq
for r in range(0,len(cds),3):   # for loop to create a range 0 to len(cds) in steps of 3 
    codon_seq = cds[r:r+3]   # slicing the codons using the range to access the value of the keys (codons) from the dictionary
    if codon_seq in codon_table:   # checks if the codons (key) is present in the dictionary and consider only those codons
        amino_seq = amino_seq+codon_table[codon_seq] # adds the amino acid residues to create the translated amino acid sequence (manual)
print("Translated sequence manually") # prints the text
print(amino_seq)  # prints the translated amino acid sequence
amino_30 =""  #initializing variable amino_30
for r in range(0,90,3):   # for loop to create a range 0 to 90 in steps of 3 which gives 30 amino acid residue sequence
    codon_30 = cds[r:r+3]   # slicing the codons using the range to access the value of the keys (codons) from the dictionary
    if codon_30 in codon_table:   # checks if the codons (key) is present in the dictionary and consider only those codons
        amino_30 = amino_30+codon_table[codon_30] # adds the amino acid 
print("Total length of translated sequence::",len(amino_seq))
print("First 30 residues:",amino_30)
```

2.(c) Compare your translation to the deposited protein from 1(c). Report whether the two are identical. If they 
are not, report the position of the first difference and what each sequence has at that position. State th 
tool you used to compare the

tool used - zip for comparisonm.

```python
with open("protein_sequence.fasta","r") as protseq:  #to open protein sequence fasta file and represent it as the variable protseq
    depo_seq =""   # initializing variable depo_seq
    for seq1 in protseq:  # for loop to read each line in protseq
        if(seq1[0]!=">"):  # checks if the first line has header symbol not to consider it instead other lines 
            depo_seq = depo_seq+seq1.strip()  # removes whitespace and adds each line and stored in variable depo_seq
print("Deposited protein sequence")  # prints the text deposited protein sequence
print(depo_seq)    # prints the deposited protein sequence
print("Length of deposited sequence:",len(depo_seq))   # calculates length of the deposited protein sequence
print("My translated sequence length:",len(amino_seq))  # calculates length of manually translated sequence
diff = 0   # initializing diff variable 
for a,d in zip(amino_seq,depo_seq):   # to compare deposited sequence and my translated sequence
    if a!=d:   # if d(deposited sequence) not equal to a (manually translated sequence) then difference is counted
        diff+=0  # diff will be update everytime a difference is found
print("Difference between my translated sequence and deposited sequence:",diff)  #print difference count
```

2.(d) Cross-check your own translation against a library function — Biopython's translate, or any equivalent. 
Confirm that the two agree, or explain exactly where and why they do not.

```python
from Bio.Seq import Seq   # to perform translation using biopython tool so import is used to access the tool
translate_seq = Seq(cds).translate()   # translating the mrna using translate() function of biopython
print(translate_seq)    # prints the translated sequence
print(len(translate_seq))   # prints length of the translated sequence
```

```python
diff1 = 0   # initializing diff1 variable 
for t,s in zip(translate_seq,amino_seq):   # to compare biopython translated sequence and my translated sequence
    if t!=s:   # if t(biopython translated sequence) not equal to a (manually translated sequence) then difference is counted
        diff1+=0  # diff1 will be update everytime a difference is found
print("Difference between my translated sequence and biopython translated sequence:",diff1)  #print difference count
```
